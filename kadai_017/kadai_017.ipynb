```python
class Human:
    def __init__(self, name, age):
        #属性を定義する
        self.name = name
        self.age = age

        #大人かどうかチェックするメソッド
    def check_adult(self):
       if self.age >= 20:
          print(self.name + ' は大人です。')
       else:
          print(self.name + ' は大人ではありません。')


# human1 = Human('ハイゼンベルク', 19)
# human2 = Human('ボーア', 20)
# human3 = Human('プランク', 21)
# humans = [human1, human2, human3]

#インスタント化する
humans = [Human('ハイゼンベルク', 19), Human('ボーア', 20), Human('プランク', 21)]

#リスト
for human in humans:
    human.check_adult()
```
